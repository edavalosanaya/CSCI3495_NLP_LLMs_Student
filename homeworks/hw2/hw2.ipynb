```python
# GIVEN. A larger generated review set, so the numbers mean something.
random.seed(1)
GOOD = ["brilliant", "great", "superb", "wonderful", "clever", "funny"]
BAD = ["dull", "boring", "awful", "weak", "terrible", "lazy"]
NOUNS = ["film", "script", "cast", "acting", "story", "ending"]
texts, labels = [], []
for _ in range(400):
    negated = random.random() < 0.25
    positive = random.random() < 0.5
    bank = GOOD if (positive != negated) else BAD
    w, n = random.choice(bank), random.choice(NOUNS)
    texts.append(f"the {n} was not {w}" if negated else f"a {w} {n}")
    labels.append("pos" if positive else "neg")

Xtr, Xte, ytr, yte = train_test_split(texts, labels, test_size=0.25,
                                      random_state=0, stratify=labels)
print(f"{len(texts)} reviews, {len(Xtr)} for training")
print("examples:", texts[:3])
```
