## Part 1. Talk to a model on your own machine

No API key, no account, no internet. The model runs in Docker on your
laptop and answers on `localhost`. If it is not running, the cell below
says so and uses a canned reply so the rest of the notebook still works.

```python
# GIVEN. Ask the local model a question.
import ollama

MODEL = "qwen2.5:0.5b"

def ask(prompt, temperature=0.7):
    """Send a prompt to the local model. Falls back to a canned reply if the
    server is not running, so this notebook always works."""
    try:
        r = ollama.chat(model=MODEL, messages=[{"role": "user", "content": prompt}],
                        options={"temperature": temperature})
        return r["message"]["content"].strip()
    except Exception as e:
        return f"[no local model running: {type(e).__name__}] canned reply: NLP is how computers work with human language."

answer = ask("In one sentence, what is natural language processing?")
print(answer)
```

## Part 2. What temperature does

`temperature` controls how much the model gambles. Low means it always
picks its favourite next word. High means it takes risks. Same prompt,
three settings, three times each.

```python
# GIVEN. The same prompt at three temperatures.
prompt = "Name one colour."

for t in (0.0, 0.8, 1.5):
    print(f"--- temperature {t} ---")
    for attempt in range(3):
        print("   ", ask(prompt, temperature=t)[:70])
```

```python
# ================== YOUR TURN 2 ==================
# Look at the three blocks above before editing anything.
#
# At temperature 0.0 the three answers should be IDENTICAL.
# At 1.5 they should differ, and may stop making sense.
#
# Now set TEMPERATURE below to the value you would use if you wanted the
# same answer every time.
#
# Expected: 0.0 gives you repeatable output. That matters later: when you
#           compare two prompts you want the model held still, so any difference
#           you see comes from the prompt and not from luck.
# ===============================================
TEMPERATURE = 1.5          # <-- change me

runs = [ask("Name one colour.", temperature=TEMPERATURE) for _ in range(3)]
for r in runs:
    print("   ", r[:70])
print()
print("all three identical:", len(set(runs)) == 1)
```
